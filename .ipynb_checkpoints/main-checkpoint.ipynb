```python
from decimal import Decimal, ROUND_DOWN

# Función para redondear al múltiplo válido de stepSize
def ajustar_step_size(cantidad, step_size_str):
    step = Decimal(step_size_str)
    cantidad_decimal = Decimal(str(cantidad))
    return float(cantidad_decimal.quantize(step, rounding=ROUND_DOWN))

# Parámetros del filtro
step_size = '0.00001000'
min_qty = 0.00001

# Obtener BTC disponible
btc_balance = next(b for b in client.get_account()['balances'] if b['asset'] == 'BTC')
btc_disponible = float(btc_balance['free'])
print(btc_disponible)
# Calcular la mitad
btc_a_vender = btc_disponible / 2

# Ajustar al múltiplo válido
btc_a_vender = ajustar_step_size(btc_a_vender, step_size)

# Verificar que sea mayor al mínimo
if btc_a_vender >= min_qty:
    orden = client.order_market_sell(symbol='BTCUSDT', quantity=btc_a_vender)
    print(f"Orden ejecutada: {orden}")
else:
    print(f"No se puede vender: cantidad ajustada ({btc_a_vender}) menor al mínimo permitido ({min_qty}).")
```

```python
from binance.client import Client
from binance.exceptions import BinanceAPIException
import os
from dotenv import load_dotenv
import numpy as np
from scipy.stats import lognorm
import time
from decimal import Decimal, ROUND_DOWN

# =================== CONFIGURACIÓN =======================
load_dotenv()
api_key = os.getenv("BINANCE_API_KEY")
api_secret = os.getenv("BINANCE_API_SECRET")
client = Client(api_key, api_secret)

PAIR = "BTCUSDT"
MONTO_USDT = 50             # Cuánto USDT usar en cada compra
CANTIDAD_BTC = 0.001        # Cuánto BTC vender
INTERVALO_ESPERA = 10       # Tiempo en segundos entre revisiones

# =================== FUNCIONES BASE =======================

def ajustar_step_size(cantidad, step_size_str):
    step = Decimal(step_size_str)
    cantidad_decimal = Decimal(str(cantidad))
    return float(cantidad_decimal.quantize(step, rounding=ROUND_DOWN))

def obtener_step_y_min_qty(symbol):
    info = client.get_symbol_info(symbol)
    filtro = next(f for f in info['filters'] if f['filterType'] == 'LOT_SIZE')
    return filtro['stepSize'], float(filtro['minQty'])

def obtener_precios_ultimos_30_dias():
    klines = client.get_historical_klines(PAIR, interval=Client.KLINE_INTERVAL_1DAY, start_str="30 days ago UTC")
    return [float(k[4]) for k in klines]  # precios de cierre

def ajustar_distribucion_lognormal(precios):
    log_precios = np.log(precios)
    shape, loc, scale = lognorm.fit(log_precios, floc=0)
    return shape, loc, scale

def calcular_umbral(shape, loc, scale, percentil):
    return np.exp(lognorm.ppf(percentil, shape, loc=loc, scale=scale))

def comprar_btc_usdt(monto_usdt):
    precio = float(client.get_symbol_ticker(symbol=PAIR)["price"])
    step_size, min_qty = obtener_step_y_min_qty(PAIR)
    cantidad = ajustar_step_size(monto_usdt / precio, step_size)

    if cantidad >= min_qty:
        try:
            orden = client.order_market_buy(symbol=PAIR, quantity=cantidad)
            print("✅ Compra ejecutada:", orden)
        except BinanceAPIException as e:
            print("❌ Error al comprar:", e)
    else:
        print(f"⚠️ Cantidad calculada ({cantidad}) menor al mínimo permitido ({min_qty}).")

def vender_btc(cantidad_btc):
    step_size, min_qty = obtener_step_y_min_qty(PAIR)
    cantidad = ajustar_step_size(cantidad_btc, step_size)

    if cantidad >= min_qty:
        try:
            orden = client.order_market_sell(symbol=PAIR, quantity=cantidad)
            print("✅ Venta ejecutada:", orden)
        except BinanceAPIException as e:
            print("❌ Error al vender:", e)
    else:
        print(f"⚠️ Cantidad calculada ({cantidad}) menor al mínimo permitido ({min_qty}).")

def decidir_y_actuar(precio_actual, p10, p90):
    print(f"💹 Precio actual: {precio_actual:.2f} | P10: {p10:.2f} | P90: {p90:.2f}")
    if precio_actual < p10:
        print("🟢 Señal de COMPRA")
        comprar_btc_usdt(MONTO_USDT)
    elif precio_actual > p90:
        print("🔴 Señal de VENTA")
        vender_btc(CANTIDAD_BTC)
    else:
        print("⏸️ Mantener posición (esperar)")

# =================== LOOP PRINCIPAL =======================

def main():
    print("🔁 Iniciando bot de trading con distribución log-normal...\n")
    while True:
        try:
            precios = obtener_precios_ultimos_30_dias()
            shape, loc, scale = ajustar_distribucion_lognormal(precios)
            p10 = calcular_umbral(shape, loc, scale, 0.10)
            p90 = calcular_umbral(shape, loc, scale, 0.90)

            precio_actual = float(client.get_symbol_ticker(symbol=PAIR)["price"])
            decidir_y_actuar(precio_actual, p10, p90)

            print(f"⌛ Esperando {INTERVALO_ESPERA} segundos...\n")
            time.sleep(INTERVALO_ESPERA)

        except Exception as e:
            print("⚠️ Error general:", e)
            time.sleep(10)  # Esperar antes de reintentar

if __name__ == "__main__":
    main()
```
